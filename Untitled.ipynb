```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("Files in Week 3 folder:")
print(os.listdir())
```

```python
iris = pd.read_csv("Iris.csv")

print(iris.head())
print("Shape:", iris.shape)
print("Columns:", iris.columns.tolist())
```

```python
# Select numerical features
X = iris[['SepalLengthCm',
          'SepalWidthCm',
          'PetalLengthCm',
          'PetalWidthCm']].values

# Standardize the data manually
mean = X.mean(axis=0)
std = X.std(axis=0)

X_scaled = (X - mean) / std

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
print("\nFirst 5 scaled rows:")
print(X_scaled[:5])
```

```python
# K-Means Clustering from scratch

def kmeans(X, k, max_iter=100, random_state=42):
    rng = np.random.default_rng(random_state)

    # Randomly select initial centers
    indices = rng.choice(len(X), size=k, replace=False)
    centers = X[indices].copy()

    for _ in range(max_iter):

        # Calculate distance from every point to every center
        distances = np.sqrt(
            ((X[:, np.newaxis, :] - centers[np.newaxis, :, :]) ** 2).sum(axis=2)
        )

        # Assign each point to nearest center
        labels = np.argmin(distances, axis=1)

        # Calculate new centers
        new_centers = np.array([
            X[labels == i].mean(axis=0) if np.any(labels == i)
            else centers[i]
            for i in range(k)
        ])

        # Stop if centers do not change
        if np.allclose(centers, new_centers):
            centers = new_centers
            break

        centers = new_centers

    return labels, centers


# Apply K-Means with K = 3
cluster_labels, cluster_centers = kmeans(X_scaled, k=3)

print("Cluster Centers:")
print(cluster_centers)

print("\nCluster Labels:")
print(cluster_labels[:20])
```

```python
# PCA: Reduce 4 dimensions to 2 dimensions

# Center the scaled data
X_centered = X_scaled - X_scaled.mean(axis=0)

# Calculate covariance matrix
cov_matrix = np.cov(X_centered, rowvar=False)

# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

# Sort eigenvalues in descending order
indices = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[indices]
eigenvectors = eigenvectors[:, indices]

# Select first 2 principal components
X_pca = X_centered @ eigenvectors[:, :2]

print("Original shape:", X_scaled.shape)
print("PCA shape:", X_pca.shape)
print("\nFirst 5 PCA values:")
print(X_pca[:5])
```

```python
# Explained Variance Ratio

explained_variance_ratio = eigenvalues / eigenvalues.sum()

print("Explained Variance Ratio:")
print(explained_variance_ratio)

print("\nFirst Principal Component:",
      explained_variance_ratio[0])

print("Second Principal Component:",
      explained_variance_ratio[1])

print("\nVariance explained by first 2 components:",
      explained_variance_ratio[:2].sum())
```

```python
# Visualize K-Means clusters using PCA

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=cluster_labels,
    cmap='viridis',
    s=50
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering of Iris Dataset")

plt.colorbar(label="Cluster")

plt.show()
```

```python
# ============================================================
# SKILL NEXIS - WEEK 3
# UNSUPERVISED LEARNING + MODEL EVALUATION
# MINI PROJECT: IRIS FLOWER CLUSTERING
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle


# ============================================================
# 1. LOAD IRIS DATASET
# ============================================================

iris = pd.read_csv("Iris.csv")

print("=" * 60)
print("IRIS DATASET")
print("=" * 60)

print(iris.head())
print("\nShape:", iris.shape)
print("\nColumns:", iris.columns.tolist())


# ============================================================
# 2. SELECT FEATURES
# ============================================================

X = iris[
    [
        'SepalLengthCm',
        'SepalWidthCm',
        'PetalLengthCm',
        'PetalWidthCm'
    ]
].values

print("\nOriginal Feature Shape:", X.shape)


# ============================================================
# 3. DATA SCALING
# ============================================================

mean = X.mean(axis=0)
std = X.std(axis=0)

X_scaled = (X - mean) / std

print("\nScaled Data Shape:", X_scaled.shape)

print("\nFirst 5 Scaled Rows:")
print(X_scaled[:5])


# ============================================================
# 4. K-MEANS CLUSTERING FROM SCRATCH
# ============================================================

def kmeans(X, k, max_iter=100, random_state=42):

    rng = np.random.default_rng(random_state)

    # Select initial cluster centers
    indices = rng.choice(
        len(X),
        size=k,
        replace=False
    )

    centers = X[indices].copy()

    for _ in range(max_iter):

        # Calculate distances
        distances = np.sqrt(
            (
                (
                    X[:, np.newaxis, :]
                    -
                    centers[np.newaxis, :, :]
                ) ** 2
            ).sum(axis=2)
        )

        # Assign points to nearest cluster
        labels = np.argmin(distances, axis=1)

        # Calculate new centers
        new_centers = np.array([
            X[labels == i].mean(axis=0)
            if np.any(labels == i)
            else centers[i]
            for i in range(k)
        ])

        # Check convergence
        if np.allclose(centers, new_centers):
            centers = new_centers
            break

        centers = new_centers

    return labels, centers


# Apply K-Means with K = 3

cluster_labels, cluster_centers = kmeans(
    X_scaled,
    k=3,
    random_state=42
)

print("\n" + "=" * 60)
print("K-MEANS CLUSTERING")
print("=" * 60)

print("\nNumber of Clusters:", 3)

print("\nCluster Centers:")
print(cluster_centers)

print("\nFirst 20 Cluster Labels:")
print(cluster_labels[:20])


# ============================================================
# 5. PCA - REDUCE 4D TO 2D
# ============================================================

X_centered = X_scaled - X_scaled.mean(axis=0)

# Covariance matrix
cov_matrix = np.cov(
    X_centered,
    rowvar=False
)

# Eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(
    cov_matrix
)

# Sort in descending order
indices = np.argsort(
    eigenvalues
)[::-1]

eigenvalues = eigenvalues[indices]

eigenvectors = eigenvectors[:, indices]

# Select first two principal components
X_pca = X_centered @ eigenvectors[:, :2]

print("\n" + "=" * 60)
print("PCA")
print("=" * 60)

print("\nOriginal Shape:", X_scaled.shape)

print("PCA Shape:", X_pca.shape)

print("\nFirst 5 PCA Values:")
print(X_pca[:5])


# ============================================================
# 6. EXPLAINED VARIANCE RATIO
# ============================================================

explained_variance_ratio = (
    eigenvalues / eigenvalues.sum()
)

print("\n" + "=" * 60)
print("EXPLAINED VARIANCE RATIO")
print("=" * 60)

print("\nExplained Variance Ratio:")
print(explained_variance_ratio)

print(
    "\nFirst Principal Component:",
    explained_variance_ratio[0]
)

print(
    "Second Principal Component:",
    explained_variance_ratio[1]
)

print(
    "\nVariance Explained by First 2 Components:",
    explained_variance_ratio[:2].sum()
)


# ============================================================
# 7. VISUALIZE K-MEANS CLUSTERS
# ============================================================

plt.figure(figsize=(8, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=cluster_labels,
    cmap='viridis',
    s=50
)

plt.xlabel("Principal Component 1")

plt.ylabel("Principal Component 2")

plt.title(
    "K-Means Clustering of Iris Dataset"
)

plt.colorbar(
    label="Cluster"
)

plt.grid(True)

plt.show()


# ============================================================
# 8. TRUE SPECIES VS PREDICTED CLUSTERS
# ============================================================

comparison = pd.crosstab(
    iris['Species'],
    cluster_labels,
    rownames=['True Species'],
    colnames=['Predicted Cluster']
)

print("\n" + "=" * 60)
print("TRUE SPECIES VS PREDICTED CLUSTERS")
print("=" * 60)

print(comparison)


# ============================================================
# 9. ELBOW METHOD
# ============================================================

inertias = []

k_values = range(1, 11)

for k in k_values:

    labels, centers = kmeans(
        X_scaled,
        k=k,
        random_state=42
    )

    # Calculate distances from points to centers
    distances = X_scaled - centers[labels]

    # Calculate inertia
    inertia = np.sum(
        distances ** 2
    )

    inertias.append(inertia)


# Plot Elbow Curve

plt.figure(figsize=(8, 5))

plt.plot(
    k_values,
    inertias,
    marker='o'
)

plt.xlabel(
    "Number of Clusters (K)"
)

plt.ylabel(
    "Inertia"
)

plt.title(
    "Elbow Method for Iris Dataset"
)

plt.xticks(
    k_values
)

plt.grid(True)

plt.show()


# ============================================================
# 10. SAVE MODEL USING PICKLE
# ============================================================

model = {
    "centers": cluster_centers,
    "mean": mean,
    "std": std,
    "k": 3
}

with open(
    "iris_kmeans_model.pkl",
    "wb"
) as file:

    pickle.dump(
        model,
        file
    )

print("\n" + "=" * 60)
print("MODEL SAVING")
print("=" * 60)

print(
    "\nModel saved successfully!"
)

print(
    "File: iris_kmeans_model.pkl"
)


# ============================================================
# 11. LOAD SAVED MODEL
# ============================================================

with open(
    "iris_kmeans_model.pkl",
    "rb"
) as file:

    loaded_model = pickle.load(
        file
    )

print("\n" + "=" * 60)
print("MODEL LOADING")
print("=" * 60)

print(
    "\nModel loaded successfully!"
)

print(
    "Saved K:",
    loaded_model["k"]
)

print(
    "\nLoaded Cluster Centers:"
)

print(
    loaded_model["centers"]
)


# ============================================================
# 12. PREDICT CLUSTER FOR NEW FLOWER
# ============================================================

new_flower = np.array(
    [[5.1, 3.5, 1.4, 0.2]]
)

# Scale new data
new_flower_scaled = (
    new_flower
    -
    loaded_model["mean"]
) / loaded_model["std"]


# Calculate distance from centers

distances = np.sqrt(
    (
        (
            new_flower_scaled
            -
            loaded_model["centers"]
        ) ** 2
    ).sum(axis=1)
)


# Find nearest cluster

predicted_cluster = np.argmin(
    distances
)

print("\n" + "=" * 60)
print("NEW FLOWER PREDICTION")
print("=" * 60)

print(
    "\nNew Flower:",
    new_flower
)

print(
    "Predicted Cluster:",
    predicted_cluster
)


# ============================================================
# 13. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("FINAL PROJECT SUMMARY")
print("=" * 60)

print(
    "\nDataset:",
    "Iris Dataset"
)

print(
    "Total Samples:",
    len(iris)
)

print(
    "Features Used:",
    4
)

print(
    "Number of Clusters:",
    3
)

print(
    "PCA Dimensions:",
    2
)

print(
    "Variance Explained by 2 PCs:",
    explained_variance_ratio[:2].sum()
)

print(
    "\nModel File:",
    "iris_kmeans_model.pkl"
)

print(
    "\nProject Completed Successfully!"
)

print("=" * 60)
```
